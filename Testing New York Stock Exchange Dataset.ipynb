```python
dffund2 = dffund.copy()
dfsec2 = dfsec.copy()
dfpsa2 = dfpsa.copy()
```

```python
dffund2 = dffund2.drop(columns=['Unnamed: 0'], errors='ignore')
```

```python
dfsec2 = dfsec2.rename(columns={'Ticker symbol': 'Ticker Symbol'})

dffund2['Ticker Symbol'] = dffund2['Ticker Symbol'].astype(str).str.strip()
dfsec2['Ticker Symbol'] = dfsec2['Ticker Symbol'].astype(str).str.strip()
```

```python
print("Duplicate tickers in securities:", dfsec2['Ticker Symbol'].duplicated().sum())
print(dfsec2[dfsec2['Ticker Symbol'].duplicated(keep=False)].sort_values('Ticker Symbol'))
```

```python
df_main_raw = dffund2.merge(dfsec2, on='Ticker Symbol', how='left', indicator=True)
```

```python
print("Fundamentals shape:", dffund2.shape)
print("Merged shape:", df_main_raw.shape)
print(df_main_raw['_merge'].value_counts())
```

```python
unmatched = df_main_raw[df_main_raw['_merge'] != 'both']['Ticker Symbol'].unique()
print("Unmatched tickers:", unmatched)
```

```python
print(df_main_raw.columns.to_list())
print(df_main_raw.head())
```

```python
print(df_main_raw['_merge'].value_counts())
```

```python
df_main_raw = df_main_raw.drop(columns=['_merge'])
```

```python
# Clean: drop columns with >50% missing values
df_clean = df_main_raw.dropna(thresh=len(df_main_raw)*0.5, axis=1)

# Filter to a single year to avoid duplicate companies per ticker
YEAR = 2015
df_clean = df_clean[df_clean['For Year'] == YEAR]

# Fix date column type
df_clean['Period Ending'] = pd.to_datetime(df_clean['Period Ending'])

print(f"Cleaned shape: {df_clean.shape}")
print(f"Remaining columns: {df_clean.columns.tolist()}")
```

```python
missing_pct = df_clean.isnull().mean().sort_values(ascending=False)
print(missing_pct.to_string())
```

```python
df_clean.info()
df_clean.shape
```

```python
df_clean['Country'] = df_clean['Address of Headquarters'].str.split(', ').str[-1]
df_clean['Country'].unique()
us_states = ['Texas', 'Virginia', 'California', 'Illinois', 'Pennsylvania',
             'Massachusetts', 'Missouri', 'Georgia', 'New York', 'Colorado', 'Louisiana', 'Washington',
             'Connecticut', 'Minnesota', 'Florida', 'Indiana', 'New Jersey', 'Tennessee', 'North Carolina',
             'Michigan', 'Nebraska', 'Ohio', 'Rhode Island', 'Oklahoma', 'Iowa', 'Delaware', 'D.C.',
             'Maryland', 'Maine', 'Utah', 'Arizona', 'Wisconsin', 'Oregon', 'Kentucky', 'Arkansas', 'Idaho',
             'South Carolina', 'Alabama', 'Nevada']
df_clean['Country'] = df_clean['Country'].str.split('[').str[0]
df_clean['Country'] = df_clean['Country'].replace('UT', 'Utah')
df_clean['Country'] = df_clean['Country'].replace('UK', 'United Kingdom')
df_clean['Country'] = df_clean['Country'].replace(us_states, 'United States')
df_clean['Country'].unique()
```

```python
df_clean.to_csv('nyse_clean_2015.csv', index=False)
```

```python
# Clean prices
dfpsa_clean = dfpsa.copy()
dfpsa_clean['date'] = pd.to_datetime(dfpsa_clean['date'])

# Filter to same year as fundamentals
dfpsa_clean = dfpsa_clean[dfpsa_clean['date'].dt.year == YEAR]

# Drop nulls (very few expected)
dfpsa_clean = dfpsa_clean.dropna()

print(f"Prices cleaned shape: {dfpsa_clean.shape}")
print(dfpsa_clean.isnull().sum())
```

```python
dfpsa_clean.to_csv('prices_split_adjusted_clean_2015.csv', index=False)
```
